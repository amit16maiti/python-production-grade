```python
## Arithmetic Operations
num_one = 5
num_two = 4
total = num_one + num_two
print(total)
diff = num_one - num_two
product = num_one * num_two
division = num_one / num_two
print(diff)
print(product)
print(division)
```

```python
##
reminder = num_one % num_two
floor_division = num_one // num_two
print(reminder)
print(floor_division)
```

```python
exp = num_one ** num_two
print(exp)
```
